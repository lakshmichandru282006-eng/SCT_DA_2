```python
import pandas as pd

df = pd.read_csv("netflix_titles.csv")

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)

display(df.head())
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
print(df.isnull().sum())
```

```python
text_columns = ["director", "cast", "country", "rating"]

for col in text_columns:
    if col in df.columns:
        df[col] = df[col].fillna("Unknown")

print("Missing values handled!")
print(df[text_columns].isnull().sum())
```

```python
df["date_added"] = pd.to_datetime(
    df["date_added"].str.strip(),
    errors="coerce"
)

# Fill missing duration with an explicit label
df["duration"] = df["duration"].fillna("Unknown")

print("Date and duration handled!")
print(df[["date_added", "duration"]].isnull().sum())
```

```python
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Duplicate rows removed:", before - after)
```

```python
df["release_year"] = pd.to_numeric(
    df["release_year"],
    errors="coerce"
).astype("Int64")

print("Updated data types:\n")
df.info()

print("\nSample cleaned records:")
display(df.head())
```

```python
df["date_added"] = df["date_added"].dt.strftime("%Y-%m-%d")

df["date_added"] = df["date_added"].fillna("Unknown")

print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nRemaining duplicate rows:", df.duplicated().sum())
```

```python
print("========== FINAL VERIFICATION ==========\n")

print("Dataset shape:\t", df.shape)

print("\nMissing values:\t")
print(df.isnull().sum())

print("\nTotal missing values:\t", df.isnull().sum().sum())

print("Duplicate rows:\t", df.duplicated().sum())

print("\nData types:\n")
print(df.dtypes)
```

```python
df.to_csv("netflix_titles_cleaned.csv", index=False)

print("Cleaned Netflix dataset exported successfully!")
print("Total rows:", len(df))
print("Total columns:", len(df.columns))
```
